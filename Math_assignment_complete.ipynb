```python
#Question 6: Force Using Newton's Second Law

#Formula:

#Force = Mass × Acceleration:

m = float(input("Enter mass in kilograms: "))

a = float(input("Enter acceleration in meters/second^2: "))

f = m * a

print ("Force =", f, "N")
```

```python
#Question 7: Compound Interest

#Formula:

#A = P(1 + r/n)^(nt)

#Where:

#A = Total Amount
#P = Principal Amount
#r = Annual Interest Rate
#n = Number of times compounded per year
#t = Time in years

P = float(input("Enter principal amount: "))
r = float(input("Enter annual interest rate (decimal): "))
n = int(input("Enter number of times compounded per year: "))
t = float(input("Enter time in years: "))

A = P * (1 + r / n) ** (n * t)

compound_interest = A - P

print("Total Amount =", A)
print("Compound Interest =", compound_interest)
```

```python
#Question 10: Kinetic Energy

#Formula:

#KE = 1/2 × m × v²

m = float(input("Enter mass in kg"))

v = float(input("Enter velocity in meters/second"))

KE = 0.5 * m * (v ** 2)

print("Kinetic Energy =", KE, "J")
```

```python
#Question 11: Quadratic Equation Roots

#Formula:

#x = (-b ± √(b² - 4ac)) / 2a

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

D = b**2 - 4*a*c


x1 = (-b + (D)) / (2*a)
x2 = (-b - (D)) / (2*a)

print("First Root =", x1)
print("Second Root =", x2)
print("The equation has no real roots.")
```

```python
#Question 13: Gravitational Force

#Formula:

#F = G × (m1 × m2) / r²

#Where:

#G = 6.674 × 10⁻¹¹

G = 6.674e-11
m1 = float(input("Enter the first object mass: "))
m2 =float(input("Enter the secound object mass: "))
r = float(input("Enter the  distance between the object in meter: "))
F = G * (m1 * m2) / r ** 2

print ("The Gravitational force =", F, "N")
```
